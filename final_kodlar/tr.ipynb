```python
# BLOK2: VERİ ÖN İŞLEME 
import pandas as pd
from tqdm import tqdm

RAW_DATA_FILE = 'openaq_raw_data_tr.csv'
PROCESSED_FILE_PATH = 'tr_air_quality_processed.csv'
COUNTRY_CODE = 'TR'
# Kirleticilerin birimleri eşitleniyor
def process_data(file_path, country_code):
    print(f"'{file_path}' dosyasından ham veri okunuyor...")
    df = pd.read_csv(file_path)
    
    print("\nBirim dönüşümü başlıyor...")
    MOLAR_MASSES = {'co': 28.01, 'no2': 46.01, 'o3': 48.00, 'so2': 64.07} # internetten alınan değerler
    MOLAR_VOLUME = 24.45
    converted_values = []
    for index, row in tqdm(df.iterrows(), total=df.shape[0], desc="Birimler Dönüştürülüyor"):
        param, unit, value = str(row['name']).lower(), str(row['unit']).lower(), row['value']
        new_value = None
        if pd.notna(value) and value >= 0:
            if unit == 'µg/m³': new_value = value
            else:
                molar_mass = MOLAR_MASSES.get(param)
                if molar_mass is not None:
                    if unit == 'ppm': new_value = (value * molar_mass * 1000) / MOLAR_VOLUME
                    elif unit == 'ppb': new_value = (value * molar_mass) / MOLAR_VOLUME
                    else: new_value = value
                else: new_value = value
        converted_values.append(new_value)
    df['value_converted'] = converted_values
    
    # Pivotlama
    print("\nVeri işleme ve pivotlama başlıyor...")
    df['from_date'] = pd.to_datetime(df['from_date'], errors='coerce')
    df.dropna(subset=['from_date', 'value_converted'], inplace=True)
    df_filtered = df[df['country'] == country_code].copy()
    
    df_pivot = df_filtered.pivot_table(index='from_date', columns='name', values='value_converted', aggfunc='mean').reset_index()
    df_pivot.columns.name = None
    df_pivot.sort_values(by='from_date', ascending=True, inplace=True)
    
    return df_pivot

try:
    processed_df = process_data(RAW_DATA_FILE, COUNTRY_CODE)
    processed_df.to_csv(PROCESSED_FILE_PATH, index=False)
    print(f"\nİşlenmiş (logaritmasız) veri '{PROCESSED_FILE_PATH}' dosyasına kaydedildi.")
    print("İşlenmiş verinin ilk 5 satırı:")
    print(processed_df.head())
except FileNotFoundError:
    print(f"Hata: '{RAW_DATA_FILE}' bulunamadı.")
```

```python
# BLOK3: FEATURE ENGINEERING
import pandas as pd
import numpy as np

PROCESSED_FILE_PATH = 'tr_air_quality_processed.csv'
ENGINEERED_FILE_PATH = 'tr_air_quality_engineered.csv'

print(f"'{PROCESSED_FILE_PATH}' dosyasından veri okunuyor...")
try:
    df = pd.read_csv(PROCESSED_FILE_PATH, parse_dates=['from_date'])
    df.set_index('from_date', inplace=True)
    print(f"Veri başarıyla yüklendi. Boyut: {df.shape}")
except FileNotFoundError:
    print(f"Hata: '{PROCESSED_FILE_PATH}' bulunamadı")
    exit()

df.sort_index(inplace=True)

# Eksik veri interpolate yöntemi ile dolduruldu
print("\nEksik veri yönetimi başlıyor...")
pollutant_cols = [col for col in df.columns if df[col].dtype in ['float64', 'int64']]
# Ülke sütunu çıkarıldı
if 'country' in pollutant_cols: pollutant_cols.remove('country')

df[pollutant_cols] = df[pollutant_cols].interpolate(method='time', limit_direction='both')
df.fillna(method='ffill', inplace=True); df.fillna(method='bfill', inplace=True)
print("Eksik veri yönetimi tamamlandı")

# Aşırı değerler kırpıldı
print("\nAşırı aykırı değerler %99.9 persentil ile kırpılıyor...")
for col in pollutant_cols:
    ust_limit = df[col].quantile(0.999)
    df[col] = df[col].clip(upper=ust_limit)
    print(f" '{col}' sütunu, üst sınır '{ust_limit:.2f}' ile kırpıldı.")

# Logaritmik dönüşüm yapıldı her bir col için hatayı minizimize etmek lazım çünkü
print("\nLogaritmik dönüşüm uygulanıyor...")
for col in pollutant_cols:
    df[col] = np.log1p(df[col])

# Türetilmiş bazı yararlı öznitelikler ekleniyor
print("\nÖznitelik mühendisliği başlıyor...")
df['year'] = df.index.year; df['month'] = df.index.month; df['day_of_week'] = df.index.dayofweek
for pollutant in pollutant_cols:
    for lag in [1, 3, 7]: df[f'{pollutant}_lag_{lag}d'] = df[pollutant].shift(lag)
    for window in [3, 7]: df[f'{pollutant}_roll_mean_{window}d'] = df[pollutant].rolling(window, closed='left').mean()
df.fillna(method='bfill', inplace=True)
print("Öznitelik mühendisliği tamamlandı")

df.to_csv(ENGINEERED_FILE_PATH, index=True)
print(f"\nTemizlenmiş veri '{ENGINEERED_FILE_PATH}' dosyasına kaydedildi")
```

```python
# BLOK3.5: VERİYİ İYİLEŞTİRME ADIMI DOĞRULTUSUNDA REGRESYONDAN CLASSİFİCTİONA DA ÇEVİRME VE KAYDETME

import pandas as pd
import numpy as np
import os

COUNTRY_CODE = 'tr' 
ENGINEERED_FILE_PATH = f'{COUNTRY_CODE}_air_quality_engineered.csv'
CATEGORIZED_FILE_PATH = f'{COUNTRY_CODE}_air_quality_categorized.csv'
ALL_TARGETS = ['co','no2','o3','pm10','pm25','so2']

# Türkiye Cumhuriyeti Çevre Şehircilik ve İklim Değişikliği Bakanlığı sitesinden alınan tabloya göre "İyi", "Orta", "Kötü" HKI eşikleri (hava kalitesi indeksleri)
HKI_ESIKLERI = {
    'so2':  [(100, 'İyi'), (250, 'Orta'), (np.inf, 'Kötü')],
    'no2':  [(100, 'İyi'), (200, 'Orta'), (np.inf, 'Kötü')],
    'co':   [(5500, 'İyi'), (10000, 'Orta'), (np.inf, 'Kötü')],
    'o3':   [(120, 'İyi'), (160, 'Orta'), (np.inf, 'Kötü')],
    'pm10': [(50, 'İyi'), (100, 'Orta'), (np.inf, 'Kötü')],
    'pm25': [(12.0, 'İyi'), (35.4, 'Orta'), (np.inf, 'Kötü')]
}

# Tek bir sayısal değeri HKI eşiklerine göre kategoriye dönüştürür.
def sayisal_degeri_kategoriye_cevir(deger, esikler):
    for sinir, kategori in esikler:
        if deger <= sinir:
            return kategori
    return esikler[-1][1]

print(f"'{ENGINEERED_FILE_PATH}' dosyası okunuyor...")
df = pd.read_csv(ENGINEERED_FILE_PATH, index_col='from_date', parse_dates=True)
df_categorized = df.copy() # Orijinal veriyi korumak için kopya 
print("Dosya okundu. Sınıflandırma işlemi başlıyor...")

available_pollutants = [p for p in ALL_TARGETS if p in df.columns]

for pollutant in available_pollutants:
    if pollutant in HKI_ESIKLERI:
        # ÖNEMLİ: HKI eşikleri orijinal değerler içindir.
        # Bu yüzden önce logaritmik dönüşümü geri aldım
        original_values = np.expm1(df[pollutant])
        esikler = HKI_ESIKLERI[pollutant]
        # Her bir değer için kategori belirlendi ve yeni bir sütuna atandı
        new_column_name = f'{pollutant}_kategori'
        df_categorized[new_column_name] = [sayisal_degeri_kategoriye_cevir(v, esikler) for v in original_values]
        print(f" -> '{new_column_name}' sütunu oluşturuldu.")

df_categorized.to_csv(CATEGORIZED_FILE_PATH)
print(f"\nSınıflandırılmış veri seti '{CATEGORIZED_FILE_PATH}' olarak kaydedildi.")
    
print("\nOluşturulan dosyanın önizlemesi:")
print(df_categorized[[col for col in df_categorized.columns if 'kategori' in col]].head())
```

```python
# BLOK4: TRAİNİNG
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split, RandomizedSearchCV, TimeSeriesSplit
from sklearn.preprocessing import RobustScaler
from sklearn.feature_selection import SelectKBest, f_regression
import joblib
import warnings
import xgboost as xgb
from sklearn.ensemble import RandomForestRegressor
from sklearn.neural_network import MLPRegressor
from sklearn.neighbors import KNeighborsRegressor

warnings.filterwarnings('ignore')

COUNTRY_CODE = 'tr' 
country_code='tr'
ENGINEERED_FILE_PATH = f'{COUNTRY_CODE}_air_quality_engineered.csv'
TARGET_POLLUTANTS = ['co', 'no2', 'o3', 'pm10', 'pm25', 'so2']
FORECAST_HORIZON = 1

# Veriyi Yükleme
df_engineered = pd.read_csv(ENGINEERED_FILE_PATH, index_col='from_date', parse_dates=True)
print(f"'{ENGINEERED_FILE_PATH}' yüklendi. Boyut: {df_engineered.shape}")

# Model Tanımları
models = {
    "xgboost": {
        "estimator": xgb.XGBRegressor(random_state=42),
        "params": {
            'n_estimators': [100, 300, 500, 700],
            'max_depth': [3, 5, 7, 9],
            'learning_rate': [0.01, 0.05, 0.1, 0.2],
            'subsample': [0.7, 0.8, 0.9, 1.0],
            'colsample_bytree': [0.7, 0.8, 0.9, 1.0]
        }
    },
    "randomforest": {
        "estimator": RandomForestRegressor(random_state=42),
        "params": {
            'n_estimators': [100, 200, 400, 600],
            'max_depth': [10, 20, 30, None],
            'min_samples_split': [2, 5, 10],
            'min_samples_leaf': [1, 2, 4],
            'max_features': ['sqrt', 'log2', 1.0]
        }
    },
    "mlp": {
        "estimator": MLPRegressor(random_state=42, max_iter=500, early_stopping=True),
        "params": {
            'hidden_layer_sizes': [(50,), (100,), (100, 50), (64, 32, 16)],
            'activation': ['relu', 'tanh'],
            'solver': ['adam'],
            'alpha': [0.0001, 0.001, 0.01, 0.05],
            'learning_rate': ['constant', 'adaptive']
        }
    },
    "knn": {
        "estimator": KNeighborsRegressor(),
        "params": {
            'n_neighbors': [3, 5, 7, 10, 15, 20],
            'weights': ['uniform', 'distance'],
            'p': [1, 2] # 1: Manhattan, 2: Euclidean
        }
    }
}

available_targets = [p for p in TARGET_POLLUTANTS if p in df_engineered.columns]

for model_name, model_info in models.items():
    for target in available_targets:
        print(f"\n>>> {country_code.upper()}: {model_name.upper()} - {target.upper()} modeli eğitiliyor...")
        
        # 1. Veri Hazırlama
        features_to_use = [col for col in df_engineered.columns if col not in TARGET_POLLUTANTS + ['country']]
        X = df_engineered[features_to_use].copy()
        y = df_engineered[[target]].shift(-FORECAST_HORIZON)
        X, y = X.iloc[:-FORECAST_HORIZON], y.iloc[:-FORECAST_HORIZON]
        X.dropna(axis=1, how='all', inplace=True); X.fillna(0, inplace=True)
        
        X_train_full, X_test, y_train_full, y_test = train_test_split(X, y, test_size=0.2, shuffle=False)
        
        # 2. Test için nesneleri oluşturma
        x_scaler = RobustScaler().fit(X_train_full)
        y_scaler = RobustScaler().fit(y_train_full)
        k_val = min(20 if model_name == 'knn' else 35, X_train_full.shape[1])
        X_train_s = pd.DataFrame(x_scaler.transform(X_train_full), columns=X_train_full.columns)
        selector = SelectKBest(score_func=f_regression, k=k_val).fit(X_train_s, y_train_full.values.ravel())
        
        joblib.dump(x_scaler, f'x_scaler_{model_name}_{target}_{country_code}.pkl')
        joblib.dump(y_scaler, f'y_scaler_{model_name}_{target}_{country_code}.pkl')
        joblib.dump(selector, f'selector_{model_name}_{target}_{country_code}.pkl')
        X_test.to_csv(f'X_test_{target}_{country_code}.csv')
        y_test.to_csv(f'y_test_{target}_{country_code}.csv')

        # 3. Hiperparametre Optimizasyonu
        X_train_hp_f = selector.transform(X_train_s)
        y_train_hp_s = y_scaler.transform(y_train_full)
        
        tscv = TimeSeriesSplit(n_splits=3)
        random_search = RandomizedSearchCV(
            estimator=model_info["estimator"], param_distributions=model_info["params"],
            n_iter=100, cv=tscv, scoring='neg_root_mean_squared_error', random_state=42, n_jobs=-1
        )
        random_search.fit(X_train_hp_f, y_train_hp_s.ravel())
        best_params = random_search.best_params_
        
        # 4. Final Modelini Tüm Eğitim Verisiyle Eğitme ve Kaydetme
        final_model = model_info["estimator"].set_params(**best_params)
        final_model.fit(X_train_hp_f, y_train_hp_s.ravel())
        joblib.dump(final_model, f'model_{model_name}_{target}_{country_code}.pkl')
        
        print(f"-> {target.upper()} için {model_name.upper()} modeli eğitildi ve kaydedildi.")
```

```python
#RNN
import pandas as pd
import numpy as np
import warnings
import time
import joblib
from sklearn.preprocessing import RobustScaler
from sklearn.model_selection import ParameterSampler, TimeSeriesSplit, train_test_split
from tqdm import tqdm
import tensorflow as tf
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, LSTM, Bidirectional, Dense, Concatenate, Dropout, BatchNormalization
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.losses import Huber
from tensorflow.keras.callbacks import EarlyStopping

warnings.filterwarnings('ignore')
tf.random.set_seed(42)
np.random.seed(42)

COUNTRY_CODE    = 'tr' # Her ülke için tüm kod aynı sadece bunu değiştirmek yeterli ('tr', 'fi', 'au')
CSV_PATH        = f'{COUNTRY_CODE}_air_quality_engineered.csv'
WINDOW_SIZE     = 7
CV_SPLITS       = 3       
RANDOM_ITERS    = 10       
CV_EPOCHS       = 30   # okuduğum makalelerde 10,50 ve 100 değerleri kullanılmıştı bende ortalama 30 olarak ayarladım 
FINAL_EPOCHS    = 100     
BATCH_SIZE      = 32      
ALL_TARGETS     = ['co','no2','o3','pm10','pm25','so2']

param_dist = {
    'units1':  [40, 64, 80], 'units2':  [20, 32, 40],
    'dropout': [0.2, 0.3, 0.4], 'lr':      [1e-3, 5e-4]
}

df = pd.read_csv(CSV_PATH, index_col='from_date', parse_dates=True)
available_targets = [t for t in ALL_TARGETS if t in df.columns]

for TARGET in available_targets:
    print("\nİŞLEM BAŞLIYOR: {COUNTRY_CODE.upper()} - {TARGET.upper()}\n")
    
    seq_feats   = [c for c in df.columns if c in ['year','month','day_of_week']+ALL_TARGETS and c != TARGET]
    stat_feats  = [c for c in df.columns if 'lag_' in c or 'roll_mean_' in c]
    all_feats   = seq_feats + stat_feats
    X_df, y_df = df[all_feats], df[[TARGET]]

    X_train_df, X_test_df, y_train_df, y_test_df = train_test_split(
        X_df, y_df, test_size=0.2, shuffle=False
    )
    
    x_scaler = RobustScaler().fit(X_train_df); y_scaler = RobustScaler().fit(y_train_df)
    
    def create_windows(X, y, w, seq_feat_count):
        Xw, Xs, yw = [], [], [];
        for i in range(len(X) - w):
            Xw.append(X[i:i+w, :seq_feat_count])
            Xs.append(X[i+w-1, seq_feat_count:])
            yw.append(y[i+w])
        return np.array(Xw), np.array(Xs), np.array(yw)

    X_train_s = x_scaler.transform(X_train_df)
    y_train_s = y_scaler.transform(y_train_df).flatten()
    Xw_tr, Xs_tr, y_tr = create_windows(X_train_s, y_train_s, WINDOW_SIZE, len(seq_feats))

    # Rnn kurulumu
    def build_model(units1, units2, dropout, lr, seq_feat_count, stat_feat_count):
        inp_seq  = Input((WINDOW_SIZE, seq_feat_count), name="sequential_input")
        inp_stat = Input((stat_feat_count,), name="statistical_input")
        x = Bidirectional(LSTM(units1, activation='tanh', return_sequences=True))(inp_seq)
        x = BatchNormalization()(x); x = Dropout(dropout)(x)
        x = Bidirectional(LSTM(units2, activation='tanh'))(x)
        x = BatchNormalization()(x); x = Dropout(dropout)(x)
        x = Concatenate()([x, inp_stat])
        x = Dense(32, activation='relu', kernel_regularizer='l2')(x)
        x = Dropout(dropout)(x)
        out = Dense(1)(x)
        m = Model([inp_seq, inp_stat], out)
        m.compile(optimizer=Adam(learning_rate=lr, clipnorm=1.0), loss=Huber(), metrics=['mae'])
        return m

    print("Hiper-parametre optimizasyonu başlıyor...")
    tscv = TimeSeriesSplit(n_splits=CV_SPLITS); best_score, best_params = np.inf, None
    param_samples = list(ParameterSampler(param_dist, n_iter=RANDOM_ITERS, random_state=42))
    for params in tqdm(param_samples, desc=f"CV Iterations for {TARGET.upper()}"):
        fold_losses = []
        for tr_idx, val_idx in tscv.split(Xw_tr):
            m = build_model(**params, seq_feat_count=len(seq_feats), stat_feat_count=len(stat_feats))
            m.fit([Xw_tr[tr_idx], Xs_tr[tr_idx]], y_tr[tr_idx],
                  epochs=CV_EPOCHS, batch_size=BATCH_SIZE, verbose=0)
            loss, _ = m.evaluate([Xw_tr[val_idx], Xs_tr[val_idx]], y_tr[val_idx], verbose=0)
            fold_losses.append(loss)
        mean_loss = np.mean(fold_losses)
        if mean_loss < best_score:
            best_score, best_params = mean_loss, params
    
    print(f"En iyi hiper-parametreler: {best_params}")

    print("Final modeli tüm eğitim verisiyle eğitiliyor...")
    final_model = build_model(**best_params, seq_feat_count=len(seq_feats), stat_feat_count=len(stat_feats))
    early_stop = EarlyStopping(monitor='val_loss', patience=15, restore_best_weights=True)
    final_model.fit([Xw_tr, Xs_tr], y_tr, epochs=FINAL_EPOCHS, batch_size=BATCH_SIZE,
                    validation_split=0.15, callbacks=[early_stop], verbose=0)
    
    # Test için gerekli  tüm dosyaları kaydetme
    file_suffix = f'rnn_{TARGET}_{COUNTRY_CODE}'    
    final_model.save(f'model_{file_suffix}.h5')
    joblib.dump(x_scaler, f'x_scaler_{file_suffix}.pkl')
    joblib.dump(y_scaler, f'y_scaler_{file_suffix}.pkl')
    joblib.dump(seq_feats, f'seq_features_{file_suffix}.pkl')
    joblib.dump(stat_feats, f'stat_features_{file_suffix}.pkl')
    X_test_df.to_csv(f'X_test_{file_suffix}.csv')
    y_test_df.to_csv(f'y_test_{file_suffix}.csv')
    
    print(f"{TARGET.upper()} için eğitim ve kaydetme tamamlandı.")
```
